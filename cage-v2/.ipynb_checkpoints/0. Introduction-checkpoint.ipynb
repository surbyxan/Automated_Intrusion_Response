# CAGE-2 Scenario with True State environment

Below is a `gym` environment that is extracted from the **evaluation.py** script, which is similar to the original evaluation script. There is a major change, where the `ChallengeWrapper` is replaced by `TrueStateChallengeWrapper`, which only differs by using `TrueStateBlueTableWrapper` instead of `BlueTableWrapper`. `TrueStateBlueTableWrapper` simply takes the all the information from the `TrueTableWrapper` and hand it over to the upper wrappers, giving the Blue agent (a.k.a the defender) the full information. See below for the differences.

```python
from pprint import pprint
import inspect

from CybORG import CybORG
from CybORG.Agents import *


path = str(inspect.getfile(CybORG))
path = path[:-10] + '/Shared/Scenarios/Scenario2.yaml'
red_agent = RedMeanderAgent
```

## Original scenario

```python
from CybORG.Agents.Wrappers import ChallengeWrapper

cyborg = CybORG(path, 'sim', agents={'Red':red_agent})
org_env = ChallengeWrapper(env=cyborg, agent_name='Blue')
obs = org_env.reset()
```

Observation of the Blue agent (Defender) in the original scenario:

### In vector form (which is used as the input for the Blue Agent evaluation)

```python
pprint(len(obs))
pprint(obs)
```

### In table form (for better visualisation)

```python
pprint(org_env.env.env.env.get_table())
```

## Simplified scenario

```python
from TrueStateChallengeWrapper import TrueStateChallengeWrapper

cyborg = CybORG(path, 'sim', agents={'Red':red_agent})
simpl_env = TrueStateChallengeWrapper(env=cyborg, agent_name='Blue')
obs = simpl_env.reset()
```

### In vector form (which is used as the input for the Blue Agent evaluation)

```python
pprint(len(obs))
pprint(obs)
```

(**Note**: Though the length is the same but the elements do not have the same meaning as in the original scenario)

### In table form (for better visualisation)

```python
pprint(simpl_env.env.env.env.get_table())
```
