```python
from pprint import pprint
import inspect
from CybORG import CybORG
from CybORG.Agents import *

from TrueStateChallengeWrapper import TrueStateChallengeWrapper

path = str(inspect.getfile(CybORG))
path = path[:-10] + '/Shared/Scenarios/Scenario2.yaml'

red_agent = RedMeanderAgent
cyborg = CybORG(path, 'sim', agents={'Red':red_agent})

env = TrueStateChallengeWrapper(env=cyborg, agent_name='Blue')

obs = env.reset()
```

# Defender actions

## Action space

As you might already know, in the deeper implementation of CybORG, the agents' actions are objects, initialised from the classes in `CybORG/Shared/Actions`:

```python
cyborg.get_action_space('Blue')
```

The above is quite complicated, but can be summarised as the following list:

```python
env.env.env.possible_actions
```

However, this is still not convenient enough—for example, what exactly is the “target” (e.g., 0x10b7622d0)? Therefore, CybORG provides EnumActionWrapper to enumerate these actions, mapping the defender’s action space into a Euclidean index space [0,144]. This convention is convenient for testing, agent training and implementation and is used in the ChallengeWrapper in the evaluation script.

```python
env.get_action_space('Blue')
```

## Action Mapping
We know have to map the integer actions into the abstract action, so that you can further do manual experimentation just by inputing the integer action to the environment. To save your time:
- the mapping follows the `possible_actions` that are printed above, top-down from 0 to 144. E.g.

```python
action = 0
env.step(action)
env.get_last_action('Blue')
```

```python
action = 144
env.step(action)
env.get_last_action('Blue')
```

- For the other host-targetting actions (such as Analyse and Remove), where you dont know what the target is for each action, look at the table view of the Blue agent. The first in the list for each action is equivalent to a target at the Defender while the last will be targetted at User4. E.g. action=2 is 'Analysing on Defender', action=3 is 'Analysing on Enterprise0'.

```python
pprint(env.env.env.env.get_table())
```

# Attacker actions

The attacker is equipped with a set of actions including: (1) Scanning a subnet (DiscoverRemoteSystems), (2) Scanning a host(DiscoverNetworkServices) , (3) Exploit a host (ExploitRemoteService), (4) Previledge escalate (PrivilegeEscalate) on an exploited a host, and (5) Impact a host (to interrupt the services running on the host).

You can learn the attacker actions by
- Studying the source code of the actions. While (1), (2), (4), and (5) are quite straight forward, (4) is more complex, where the attacker choose the type of vulnerabilities to exploit (BlueKeep, EternalBlue,FTPDirectoryTraversal...) and with varied result.
- Doing manual experiments with the `KeyboardAgent` as follows.

```python
from IPython.display import clear_output

from CybORG.Agents import KeyboardAgent, SleepAgent
from CybORG.Agents.Wrappers import RedTableWrapper
red_agent = KeyboardAgent()
cyborg = CybORG(path, 'sim', agents={'Blue':SleepAgent})
env = RedTableWrapper(env=cyborg) # We now act as an attacker with an idle defender
results = env.reset('Red')
observation = results.observation
action_space = results.action_space
```

```python
for i in range(100):
    print("True state of the system")
    print(env.env.get_table())
    print("Observation of the attacker")
    action = red_agent.get_action(observation=observation, action_space=action_space)
    results = env.step(action=action, agent='Red')
    observation = results.observation
    action_space = results.action_space
    #print(obs)

    clear_output(wait=False)
```
