```python
from pprint import pprint
import inspect
from CybORG import CybORG
from IPython.display import clear_output

from TrueStateChallengeWrapper import TrueStateChallengeWrapper

path = str(inspect.getfile(CybORG))
path = path[:-10] + '/Shared/Scenarios/Scenario2.yaml'

```

In the scenario, you will develop policies to defender the system against three types of attacker:
- `SleepAgent`: Bassically this means no attacker in the system
- `RedMeanderAgent`: The attacker has more randomness in its attack path and can be less predictable
- `B_lineAgent`: The attacker is assumed to have knowledge about the network topology and goes straight forward to the target host, i.e. `Op_Server0`

Your task now is to learn the behaviour of each attacker strategy, in two ways:
- Study the implementation of the attacker through their source codes
- Doing manual experimentation as follows

# Manual experimentation
You can run manual steps with action 0 (Sleep and do nothing), watch the system states (table form, vector form) and see how it evolves.

## RedMeanderAgent

```python
from CybORG.Agents.SimpleAgents.Meander import RedMeanderAgent
red_agent = RedMeanderAgent
cyborg = CybORG(path, 'sim', agents={'Red':red_agent})
env = TrueStateChallengeWrapper(env=cyborg, agent_name='Blue')
obs = env.reset()
```

```python
for i in range(100):
    clear_output(wait=True)
    a= input("Press Enter to continue...")
    if a:
        break
    action = 0
    env.step(action)
    print(env.env.env.env.get_table())
    red_last_action = env.env.env.env.env.env.environment_controller.get_last_action('Red')
    print(f" Attacker action: {red_last_action}")
    
```
