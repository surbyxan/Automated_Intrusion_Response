As you have learned about the attacker actions and defender actions, now you should learn about the dynamic of the system, i.e., how the system state changes given the attacker and the defender actions. Unfortunately, I could not find anyway to manually choose both the attacker action and the defender action in the same time step. Therefore, the best way is to use `KeyboardAgent` for the defender and try manually choose the action against one of the two type of attacker strategies that are used in the evaluation: `RedMeanderAgent` or `B_LineAgent`. Also, please read the source code of the actions as well.

A little information to save you some time: In a time step, though it is said that the attacker and the defender simultaneously perform actions, CybORG actually applies the defender action first, then applies the attacker action.

# Manual experimentation
You can run manual steps with action 0 (Sleep and do nothing), watch the system states (table form, vector form) and see how it evolves.

```python
from CybORG.Agents.SimpleAgents.Meander import RedMeanderAgent
from IPython.display import clear_output

from CybORG.Agents import KeyboardAgent
from TrueStateBlueTableWrapper import TrueStateBlueTableWrapper
defender = KeyboardAgent()
cyborg = CybORG(path, 'sim', agents={'Red':RedMeanderAgent})
env = TrueStateBlueTableWrapper(env=cyborg) # We now act as the defender
results = env.reset('Blue')
observation = results.observation
action_space = results.action_space
```

```python
for i in range(100):
    print("True state of the system / Defender observation")
    print(env.env.get_table())
    action = defender.get_action(observation=observation, action_space=action_space)
    results = env.step(action=action, agent='Blue')
    observation = results.observation
    action_space = results.action_space
    #print(obs)

    clear_output(wait=False)
    
```
