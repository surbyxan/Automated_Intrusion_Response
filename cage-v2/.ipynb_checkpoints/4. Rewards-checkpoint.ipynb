```python
from pprint import pprint
import inspect
from CybORG import CybORG
from IPython.display import clear_output

from TrueStateChallengeWrapper import TrueStateChallengeWrapper

path = str(inspect.getfile(CybORG))
path = path[:-10] + '/Shared/Scenarios/Scenario2.yaml'
```

For each time step, the defender receives a non-positive reward, based on the system state and the defender action. The objective of the defender (stopping the attack and maintaining services in the system) is presented by the goal to **maximise** the cumulative reward over a number of time steps. To learn how the reward is calculated, it is best to study CAGE-2 documentation and to manually run the game and observe the states and the rewards as follows. 

(You can also read the source code to see how it is defined, but it's complicated and takes time so I would not recommend this way).

# Manual experimentation
Tip: Run with action=0 (Sleep) first and observe how the rewards relate to the states of the system. Then you can try different actions to see if they affect the reward, beside the states. Look at the reward breakdown if needed.

```python
from CybORG.Agents.SimpleAgents.Meander import RedMeanderAgent
from IPython.display import clear_output

cyborg = CybORG(path, 'sim', agents={'Red':RedMeanderAgent})
env = TrueStateChallengeWrapper(env=cyborg, agent_name='Blue')
result = env.reset()
```

```python
for i in range(100):
    clear_output(wait=True)
    a = input("Press enter the action (0-144) or nothing to quit")
    if not a:
        break
    action = int(a)
    observation, reward, done, info = env.step(action)
    print(env.env.env.env.get_table())
    print(f"Defender Reward: {reward}")
    print(f"Reward break down")
    pprint(env.get_reward_breakdown("Blue"))
    
```
