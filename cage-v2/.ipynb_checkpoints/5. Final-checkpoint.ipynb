# Final

The objective of the project, is to find (good) defender strategy (as a function) that takes an input (defender observation) to an action. As provided in `evaluation.py`, the input is defined **by me** as a vector (as you saw from the Introduction notebook), formulated in the `TrueStateBlueTableWrapper`. It is my simple presentation of the state of the system, based on the table view of the system. It is fine an acceptable if you use it in the implementation, however, it might not be fully correct in presenting all states of the system. There are cases that are hard to resimulate and detect. For examples:

If you run the following simulation, choosing action against `RedMeanderAgent` attacker. You might find that if the attacker perform `Exploit` action on a host and chooses its sub-action (exploitation) as `SSHBruteForce`, the state of the system is defined as `Access:User`, like many other sub-actions. In the previous experimentations, you might have found that `Remove` action can convert this access to `None` (i.e., revoke the attacker out of the host), however, it does not revoke the access if it was `SSHBruteForce`ed. 

```python
from pprint import pprint
import inspect
from CybORG import CybORG
from IPython.display import clear_output

from CybORG.Agents.SimpleAgents.Meander import RedMeanderAgent
from TrueStateChallengeWrapper import TrueStateChallengeWrapper

path = str(inspect.getfile(CybORG))
path = path[:-10] + '/Shared/Scenarios/Scenario2.yaml'

cyborg = CybORG(path, 'sim', agents={'Red':RedMeanderAgent})
env = TrueStateChallengeWrapper(env=cyborg, agent_name='Blue')
result = env.reset()
```

```python
for i in range(100):
    clear_output(wait=True)
    a = input("Press enter the action (0-144) or nothing to quit")
    if not a:
        break
    action = int(a)
    observation, reward, done, info = env.step(action)
    print(env.env.env.env.get_table())
    # Inspect actions:
    defender_last_action = env.env.env.env.env.env.environment_controller.get_last_action('Blue')
    print(f"Defender action: {defender_last_action}")
    attacker_last_action = env.env.env.env.env.env.environment_controller.get_last_action('Red')
    print(f" Attacker action: {attacker_last_action}")
    if attacker_last_action.__class__.__name__ == "ExploitRemoteService":
        print(getattr(attacker_last_action, 'sub_action', None))
    
```
